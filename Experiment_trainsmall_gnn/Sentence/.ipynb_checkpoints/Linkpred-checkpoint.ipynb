```python
import torch
import torch_geometric
import pickle
import torch_geometric.transforms as T
import torch_sparse
```

```python
PTH = r'./picklefiles'

with open(f'{PTH}//graph_trainsmall_linkpred.pkl', 'rb') as f:
    graph = pickle.load(f)
```

```python
graph
```

```python
graph = T.ToUndirected()(graph)
```

```python
graph
```

```python
transform = T.RandomLinkSplit(
    disjoint_train_ratio = 0.7,   
    add_negative_train_samples = True,  
    edge_types = ("target_entity", "linked_to", "aspect"),
    rev_edge_types = ("aspect", "rev_linked_to", "target_entity")
)
```

```python
train_data, val_data, test_data = transform(graph)
```

```python
train_data
```

```python
from torch_geometric.loader import LinkNeighborLoader
edge_label_index = train_data["target_entity", "linked_to", "aspect"].edge_label_index
edge_label = train_data["target_entity", "linked_to", "aspect"].edge_label

train_loader = LinkNeighborLoader(
    data = train_data,  
    num_neighbors= [20, 10], 
    edge_label_index=(("target_entity", "linked_to", "aspect"), edge_label_index),
    edge_label=edge_label,
    batch_size=128,
    shuffle=True,
)
```

```python
sampled_data = next(iter(train_loader))
```

```python
class GNNEncoder(torch.nn.Module):
    def __init__(self, hidden_channels, out_channels):
        super().__init__()
        self.conv1 = SAGEConv((-1, -1), hidden_channels)
        self.conv2 = SAGEConv((-1, -1), out_channels) #Add more layers or dropouts or BNs


    def forward(self, x, edge_index):
        x = self.conv1(x, edge_index).relu()
        x = self.conv2(x, edge_index)
        return x

class EdgeDecoder(torch.nn.Module):
  def __init__(self, hidden_channels):
    super().__init__()
    self.lin1 = torch.nn.Linear(2 * hidden_channels, hidden_channels)
    self.lin2 = torch.nn.Linear(hidden_channels, 1)

  def forward(self, z_dict, edge_label_index):
    row, col = edge_label_index
    z = torch.cat([z_dict['target_entity'][row], z_dict['aspect'][col]], dim=-1)
    z = self.lin1(z).relu()
    z = self.lin2(z)
    return z.view(-1)


class Model(torch.nn.Module):
    def __init__(self, hidden_channels):
        super().__init__()
        self.encoder = GNNEncoder(hidden_channels, hidden_channels)
        self.encoder = to_hetero(self.encoder, graph.metadata(), aggr='mean')
        self.decoder = EdgeDecoder(hidden_channels)

    def forward(self, x_dict, edge_index_dict, edge_label_index):
        z_dict = self.encoder(x_dict, edge_index_dict)
        return self.decoder(z_dict, edge_label_index)

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

model = Model(hidden_channels=150).to(device)
print(model)
```

```python
import tqdm
import torch.nn.functional as F

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Device: '{device}'")

model = model.to(device)
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)

for epoch in range(1, 101):
    total_loss = total_examples = 0
    for sampled_data in tqdm.tqdm(train_loader):
        optimizer.zero_grad()
        sampled_data = sampled_data.to(device)
        pred = model(sampled_data.x_dict, sampled_data.edge_index_dict,
                 sampled_data['target_entity', 'aspect'].edge_label_index)
        edge_label = sampled_data['target_entity', 'aspect'].edge_label
        edge_label = edge_label.to(device)
        loss = F.binary_cross_entropy_with_logits(pred, edge_label)
        loss.backward()
        optimizer.step()
        total_loss += float(loss) * pred.numel()
        total_examples += pred.numel()
    print(f"Epoch: {epoch:03d}, Loss: {total_loss / total_examples:.4f}")
```

```python
from torch_geometric.loader import LinkNeighborLoader
edge_label_index = test_data["target_entity", "linked_to", "aspect"].edge_label_index
#edge_label_index = edge_label_index.reshape([2, 5390])
edge_label = test_data["target_entity", "linked_to", "aspect"].edge_label


test_loader = LinkNeighborLoader(
    data = test_data.contiguous(),
    edge_label_index = (('target_entity', 'linked_to', 'aspect'), edge_label_index),
    num_neighbors = [20, 10],  # TODO
    edge_label=edge_label,
    batch_size=128,
    shuffle=True,
)
```

```python
from sklearn.metrics import f1_score

preds = []
ground_truths = []
f1_tot = 0
count = 0
for sampled_data in tqdm.tqdm(test_loader):
    with torch.no_grad():
        model.eval()
        sampled_data = sampled_data.to(device)
        preds = model(sampled_data.x_dict, sampled_data.edge_index_dict,
                 sampled_data['target_entity', 'aspect'].edge_label_index)
        ground_truths = sampled_data['target_entity', 'aspect'].edge_label
        pred = torch.sigmoid(preds)
        pred = pred.to('cpu').numpy()
        pred_y = [1 if el > 0.5 else 0 for el in pred]
        ground_truths = ground_truths.to('cpu').numpy()
        ground_truths = ground_truths.astype('int')
        f1_tot += f1_score(ground_truths, pred_y)
        count += 1
        
print(f"F1_score: {f1_tot/count:.4f}")
```
