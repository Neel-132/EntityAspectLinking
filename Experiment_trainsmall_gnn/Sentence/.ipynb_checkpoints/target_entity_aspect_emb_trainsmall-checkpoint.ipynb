```python
import embedding
from transformers import BertTokenizer
from tqdm import tqdm
import torch
import pickle
```

```python
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
```

```python
PTH = r'.\picklefiles'
with open(f'{PTH}\\eal_gnn_trainsmall.pkl', 'rb') as eal:
    data = pickle.load(eal)
eal.close()

ent = [data[i][0] for i in range(len(data))]

asp = [data[i][1] for i in range(len(data))]
```

```python
tokenizer = BertTokenizer.from_pretrained("bert-base-uncased")
target_emb = torch.zeros(len(ent), 150)
target_emb.to(device)
pretrained = 'bert-base-uncased'
ent_emb = embedding.EntityEmbedding(pretrained = pretrained, device = device)
for i in range(len(ent)):
    entity_word = ent[i]['target_entity']
    tokens = tokenizer.tokenize(entity_word)
    input_ids = tokenizer.convert_tokens_to_ids(tokens)
    input_ids = torch.tensor(input_ids).unsqueeze(0).to(device)
    target_emb[i] = ent_emb(input_ids, dim = 150)
```

```python
with open(f'{PTH}\\targetentemb_sent_gnn_trainsmall.pkl', 'wb') as f:
    pickle.dump(target_emb, f)
print("Dumped")
f.close()
```

```python
key = {}
tokenizer = BertTokenizer.from_pretrained("bert-base-uncased")

for ind in tqdm(range(len(asp))):
    aspect = asp[ind]['true_aspect']
    true_aspect_id = asp[ind]['true_aspect_id']
    tokens = tokenizer.tokenize(aspect)
    input_ids = tokenizer.convert_tokens_to_ids(tokens)
    input_ids = torch.tensor(input_ids).unsqueeze(0).to(device)
    key['T' + true_aspect_id] = ent_emb(input_ids, dim = 150)

    for element in asp[ind]['candidate_aspects']:
        if len(element['aspect_name']) == 0:
            continue
        if element['aspect_id'] != true_aspect_id:
            tokens = tokenizer.tokenize(element['aspect_name'])
            input_ids = tokenizer.convert_tokens_to_ids(tokens)
            input_ids = torch.tensor(input_ids).unsqueeze(0).to(device)
            key[element['aspect_id']] = ent_emb(input_ids, dim = 150)
            
```

```python
aspemb = torch.zeros(len(key.keys()), 150).to(device)
i = 0
for el in key:
    aspemb[i] = key[el]
    i += 1
```

```python
with open(f'{PTH}\\aspemb_sent_gnn_trainsmall.pkl', 'wb') as f:
    pickle.dump(aspemb, f)
print("Dumped")
f.close()
```

```python
with open(f'{PTH}\\aspembdict_sent_gnn_trainsmall.pkl', 'wb') as f:
    pickle.dump(key, f)
print("Dumped")
f.close()
```
