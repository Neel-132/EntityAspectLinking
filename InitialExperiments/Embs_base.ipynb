```python
def decode(str):
    return urllib.parse.unquote(str)
```

```python
import re
def preprocess(text):
    text = text.lower()
    text = re.sub(r"[^\w\s]","",text)
    text = re.sub(r"(?i)\b((?:https?://|www\d{0,3}[.]|[a-z0-9.\-]+[.][a-z]{2,4}/)(?:[^\s()<>]+|\(([^\s()<>]+|(\([^\s()<>]+\)))*\))+(?:\(([^\s()<>]+|(\([^\s()<>]+\)))*\)|[^\s`!()\[\]{};:'\".,<>?«»“”‘’]))","",text)
    text = re.sub("<(\"[^\"]*\"|'[^']*'|[^'\">])*>","",text)    
    return text
```

```python
txtemb = embedding.TextEmbedding()
```

```python
paraemb = torch.zeros(len(ent),384)
for i in range(len(ent)):
    token = ent[i]["paragraph"].split(".")
    temp = torch.zeros(len(token),384)
    for j in range(len(token)):
        sent = preprocess(token[j])
        output = txtemb(sent)
        temp[j]= output
    #print(temp.shape)
    paraemb[i] = torch.mean(temp,0)
```

```python
tokenizer = BertTokenizer.from_pretrained("bert-base-uncased")
target_emb = torch.zeros(len(ent), 100)
pretrained = 'bert-base-uncased'
ent_emb = embedding.EntityEmbedding(pretrained = pretrained)
for i in range(len(ent)):
    entity_word = ent[i]['target_entity']
    tokens = tokenizer.tokenize(entity_word)
    input_ids = tokenizer.convert_tokens_to_ids(tokens)
    input_ids = torch.tensor(input_ids).unsqueeze(0) 
    target_emb[i] = ent_emb(input_ids)
```
