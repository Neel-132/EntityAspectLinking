```python
import pickle
import torch
with open(r"C:\Users\Neela\Documents\GitHub\EntityAspectLinking\picklefiles\final_eal.pkl", 'rb') as eal:
    data = pickle.load(eal)

ent = [data[i][0] for i in range(len(data))]

asp = [data[i][1] for i in range(len(data))]
```

```python
import re
def preprocess(text):
    text = text.lower()
    text = re.sub(r"[^\w\s]","",text)
    text = re.sub(r"(?i)\b((?:https?://|www\d{0,3}[.]|[a-z0-9.\-]+[.][a-z]{2,4}/)(?:[^\s()<>]+|\(([^\s()<>]+|(\([^\s()<>]+\)))*\))+(?:\(([^\s()<>]+|(\([^\s()<>]+\)))*\)|[^\s`!()\[\]{};:'\".,<>?«»“”‘’]))","",text)
    text = re.sub("<(\"[^\"]*\"|'[^']*'|[^'\">])*>","",text)
    return text
    
```

```python
txtemb = embedding.TextEmbedding()
paraemb = torch.zeros(len(ent),384)
for i in range(len(ent)):
    token = ent[i]["paragraph"].split(".")
    temp = torch.zeros(len(token),384)
    for j in range(len(token)):
        sent = preprocess(token[j])
        output = txtemb(sent)
        temp[j]= output
    #print(temp.shape)
    paraemb[i] = torch.mean(temp,0)
    
```

```python
su = 0
for el in asp:
   su += len(el['candidate_aspects'])

su
```

```python
aspemb = torch.zeros(su,384)
for i in range(len(asp)):
    #print(len(asp[i]["candidate_aspects"]))
    for candasp in asp[i]["candidate_aspects"]:
        token = candasp["content"].split(".")
        temp = torch.zeros(len(token),384)
        for j in range(len(token)):
            sent = preprocess(token[j])
            output = txtemb(sent)
            temp[j]= output
        #print(temp.shape)
        aspemb[i] = torch.mean(temp,0)
```
