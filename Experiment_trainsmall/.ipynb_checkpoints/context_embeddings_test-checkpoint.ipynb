```python
import embedding
from transformers import BertTokenizer
from tqdm import tqdm
import torch
```

```python
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
```

```python
import pickle
import torch
with open("C:\\Users\\Neela\\Documents\\GitHub\\EntityAspectLinking\\Experiment_trainsmall\\picklefiles\\eal_test.pkl", 'rb') as eal:
    data = pickle.load(eal)

ent = [data[i][0] for i in range(len(data))]

asp = [data[i][1] for i in range(len(data))]
```

```python
import re
def preprocess(text):
    text = text.lower()
    text = re.sub(r"[^\w\s]","",text)
    text = re.sub(r"(?i)\b((?:https?://|www\d{0,3}[.]|[a-z0-9.\-]+[.][a-z]{2,4}/)(?:[^\s()<>]+|\(([^\s()<>]+|(\([^\s()<>]+\)))*\))+(?:\(([^\s()<>]+|(\([^\s()<>]+\)))*\)|[^\s`!()\[\]{};:'\".,<>?«»“”‘’]))","",text)
    text = re.sub("<(\"[^\"]*\"|'[^']*'|[^'\">])*>","",text)    
    return text
```

```python
txtemb = embedding.TextEmbedding(device = device)
```

```python
paraemb = torch.zeros(len(ent),384).to(device)
for i in tqdm(range(len(ent))):
    token = ent[i]["paragraph"].split(".")
    temp = torch.zeros(len(token),384).to(device)
    for j in range(len(token)):
        sent = preprocess(token[j])
        output = txtemb(sent)
        temp[j]= output
    #print(temp.shape)
    paraemb[i] = torch.mean(temp,0)
```

```python
with open('C:\\Users\\Neela\\Documents\\GitHub\\EntityAspectLinking\\Experiment_trainsmall\\picklefiles\\contextemb_test.pkl', 'wb') as f:
    pickle.dump(paraemb, f)
print("Dumped")
f.close()
```
