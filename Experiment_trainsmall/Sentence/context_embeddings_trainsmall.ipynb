```python
import embedding
from transformers import BertTokenizer
from tqdm import tqdm
import torch
```

```python
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
```

```python
import pickle
import torch
with open(r"C:\Users\SOHAM\Desktop\Entity_Aspect_Linking\EntityAspectLinking\Experiment_trainsmall\Sentence\picklefiles\eal_trainsmall.pkl", 'rb') as eal:
    data = pickle.load(eal)

ent = [data[i][0] for i in range(len(data))]

asp = [data[i][1] for i in range(len(data))]
```

```python
import re
def preprocess(text):
    text = text.lower()
    text = re.sub(r"[^\w\s]","",text)
    text = re.sub(r"(?i)\b((?:https?://|www\d{0,3}[.]|[a-z0-9.\-]+[.][a-z]{2,4}/)(?:[^\s()<>]+|\(([^\s()<>]+|(\([^\s()<>]+\)))*\))+(?:\(([^\s()<>]+|(\([^\s()<>]+\)))*\)|[^\s`!()\[\]{};:'\".,<>?«»“”‘’]))","",text)
    text = re.sub("<(\"[^\"]*\"|'[^']*'|[^'\">])*>","",text)    
    return text
```

```python
txtemb = embedding.TextEmbedding(device = device)
```

```python
sentemb = torch.zeros(len(ent),384).to(device)
for i in tqdm(range(len(ent))):
    token = ent[i]["sentence"]
    sent = preprocess(token)
    output = txtemb(sent)    
    sentemb[i] = output
```

```python
with open(r"C:\Users\SOHAM\Desktop\Entity_Aspect_Linking\EntityAspectLinking\Experiment_trainsmall\Sentence\picklefiles\sentemb_trainsmall.pkl", 'wb') as f:
    pickle.dump(sentemb, f)
print("Dumped")
f.close()
```
