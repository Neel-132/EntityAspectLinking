```python
def make_csv(df, filename):
    df['trec_format_1'] = [0] * df.shape[0]
    df['comment'] = ['run'] * df.shape[0]
    del df['IsTrue']
    cols = df.columns.tolist()
    x = 'trec_format_1'
    cols.remove(x)
    cols = [cols[0]] + [x] + cols[1 : ]
    df = df[cols]
    df.to_csv(f'{DUMP}\\{filename}', sep = ' ',header = False, index = False)
    print('File saved successfully')
```

```python
make_csv(dnn, 'train_small_para--dnn.run')
```
